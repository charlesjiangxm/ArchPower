# Read NPY

```python
import numpy as np
import os

DATA_DIR = "dataset/component_feature"

# Load all .npy files into a dictionary
components = {}
for fname in sorted(os.listdir(DATA_DIR)):
    if fname.endswith(".npy"):
        name = fname.replace(".npy", "")
        components[name] = np.load(os.path.join(DATA_DIR, fname))

# Print summary
for name, data in components.items():
    print(f"{name:10s}  shape={str(data.shape):12s}  dtype={data.dtype}")
```

```python
# Quick look at each component's statistics
for name, data in components.items():
    print(f"\n--- {name} ({data.shape[1]} features) ---")
    print(f"  min:  {data.min(axis=0)}")
    print(f"  max:  {data.max(axis=0)}")
    print(f"  mean: {data.mean(axis=0)}")
```
